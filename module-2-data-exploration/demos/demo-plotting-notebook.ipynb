```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
fao_world = pd.read_csv("faostat-aglanduse-world.csv")
fao_world
```

```python
plt.plot(fao_world['Year'], fao_world['Value'])
```

```python
fig, ax = plt.subplots(figsize=(15, 6), nrows=1, ncols=2)

## Matplotlib 
ax[0].plot(fao_world['Year'], fao_world['Value'], color='blue', marker='o', linestyle='-', label='This is a label')

## Pandas
fao_world.plot.line(x='Year', y='Value', ax=ax[1])


ax[0].set_title('Made by Matplotlib')
ax[1].set_title('Made by Pandas')


ax[0].set_xlabel('Year')
ax[0].set_ylabel('Value')

## Pandas
ax[1].set_xlabel('Year')
ax[1].set_ylabel('Value')

ax[0].legend()

fig.show()
```
